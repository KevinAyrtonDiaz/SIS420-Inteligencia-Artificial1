```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# 1. CARGA DE DATOS
# Usamos exactamente los nombres que tienen tus carpetas y archivos
ruta_archivo = '9. Meningitis Dataset with Missing Values - Kaggle/mening missing 12.csv'
df = pd.read_csv(ruta_archivo)

print("--- EXPLORACIÓN INICIAL ---")
print("Dimensiones originales (Filas, Columnas):", df.shape)

# 2. LIMPIEZA DE DATOS (Tratamiento de Nulos)
print("\n--- VALORES NULOS ENCONTRADOS ---")
nulos_previos = df.isnull().sum().sum()
print(f"Total de datos faltantes antes: {nulos_previos}")

for col in df.columns:
    if df[col].isnull().any(): 
        if df[col].dtype == 'object': 
            df[col].fillna(df[col].mode()[0], inplace=True)
        else: 
            df[col].fillna(df[col].median(), inplace=True)

# 3. TRANSFORMACIÓN (Encoding)
df_preparado = pd.get_dummies(df, drop_first=True)

print("\n--- RESULTADO FINAL ---")
print("Dimensiones después de preparar:", df_preparado.shape)
print(f"Total de datos faltantes ahora: {df_preparado.isnull().sum().sum()}")

# 4. GUARDAR EL RESULTADO
ruta_guardado = '9. Meningitis Dataset with Missing Values - Kaggle/Meningitis_Preparado.csv'
df_preparado.to_csv(ruta_guardado, index=False)
print(f"\n¡Listo! Dataset guardado en: {ruta_guardado}")
```
